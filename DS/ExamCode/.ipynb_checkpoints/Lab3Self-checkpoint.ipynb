```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler 
from sklearn.metrics import mean_squared_error, root_mean_squared_error, r2_score
from sklearn.pipeline import Pipeline
from sklearn.linear_model import RidgeCV
```

```python
# Task-
# X 2D and Y prep
# X train and X test prediction
# Correlation
# Standard Scaler
# LinearModel
# Ridge 
# Lasso
# coeff, aplha, pipeline use dekh lena
# Then Score r2 root means, yeh sab
```

```python
df = pd.read_csv("Credit.csv")
```

```python
df.dtypes
```

```python
df.head()
```

```python
numerical_columns = df.select_dtypes(include = ['number'])
```

```python
numerical_columns.columns
```

```python
numerical_columns.drop(columns=['ID'],inplace=True)
```

```python
numerical_columns.columns
```

```python
# Correlation between all
#  Pandas way 
corr = numerical_columns.corr()
corr
```

```python
X = numerical_columns.to_numpy()
```

```python
corr = np.corrcoef(X.T)
# All X features ek line mein chahta hamesha
corr
```

```python
X = numerical_columns.drop(columns = ['Limit'])
y= numerical_columns['Limit']
```

```python
X.shape
```

```python
y.shape
```

```python
scaler = StandardScaler()
```

```python
X_scaled = scaler.fit(X)
```

```python
X_scaled.scale_
# Each column ka separate scale and mean
X_scaled.mean_
```

```python
X_scaled = scaler.transform(X)
```

```python
np.isnan(X_scaled).sum()
```

```python
type(y)
```

```python
y = y.to_numpy()
```

```python
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.4, random_state=0)
```

```python
X_train.shape
X_test.shape
```

```python
# Linear Model OLS Based Parameters aayenge
linearReg= LinearRegression()
linearReg.fit(X_train,y_train)
y_test_pred= linearReg.predict(X_test)
print("Coef",linearReg.coef_)
print("intercept",linearReg.intercept_)
# After prediction accuracies
print("R2score",r2_score(y_test, y_test_pred))
print("RMSE",root_mean_squared_error(y_test, y_test_pred))
```

```python
# without scaled x using pipeline
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.4, random_state=0)
```

```python
# Pipeline to use Ridge and Lassso
ridge = Pipeline([('scaler', StandardScaler()), ('ridge', RidgeCV())])
ridge.fit(X_train,y_train)
y_test_pred= ridge.predict(X_test)
# print(dir(ridge))
my_ridge= ridge.named_steps["ridge"]
print("Coef",my_ridge.coef_)
print("intercept",my_ridge.intercept_)
print("alpharidge",my_ridge.alpha_)
my_scaler = ridge.named_steps["scaler"]
print("scales",my_scaler.scale_ )
# After prediction accuracies
print("R2score",r2_score(y_test, y_test_pred))
print("RMSE",root_mean_squared_error(y_test, y_test_pred))
```
